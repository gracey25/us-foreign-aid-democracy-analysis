```python
import pandas as pd

# 1. Load local raw V-Dem file (uploaded to Colab files sidebar)
vdem_raw_path = "V-Dem-CY-Core-v16.csv"

df = pd.read_csv(vdem_raw_path)

# 2. Select required variables
columns_to_keep = [
    "country_name",
    "country_text_id",
    "year",
    "v2x_polyarchy",
    "v2x_libdem",
    "v2x_cspart"
]

df_subset = df[columns_to_keep]

# 3. Filter for observation years 1999 through 2023
df_subset = df_subset[(df_subset["year"] >= 1999) & (df_subset["year"] <= 2023)]

# 4. Save the trimmed dataset locally in Colab
df_subset.to_csv("vdem_1999_2023_subset.csv", index=False)

print(f"Original shape: {df.shape}")
print(f"Trimmed shape: {df_subset.shape}")
print("Pre-subsetting complete!")
```
